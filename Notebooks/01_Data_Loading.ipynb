<h1>Feature Engineering & Data Preprocessing

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

<h2>Categorical Encoding

```python
df_encoded = pd.get_dummies(
    df,
    columns=["country", "gender"],
    drop_first=True,
    dtype=int
)

df_encoded.head()
```

```python
df_encoded.columns
```

```python
df_encoded.head()
```

<h2>Features (X), Target (y)

```python
X = df_encoded.drop(["customer_id", "churn"], axis=1)

y = df_encoded["churn"]
```

```python
print(X.shape)

print(y.shape)
```

<h2>Train-Test Split<h2>

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

<h2>Feature Scaling<h2>

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
print(X_train[:5])
```

<h1>Logistic Regression<h1>

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:20])
```

<h3>Model Evaluation

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy*100:.2f}%")
```

<h3>Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

<h3>Classification Report

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

<h1>Decision Tree<h1>

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_model.fit(X_train, y_train)

```

```python
dt_pred = dt_model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, dt_pred)*100,'\n')

print(confusion_matrix(y_test, dt_pred),'\n')

print(classification_report(y_test, dt_pred))
```

<h4>Observation:</h4>

The Decision Tree model achieved lower overall accuracy than Logistic Regression but significantly improved churn detection. Recall increased from 19% to 51%, allowing the model to identify more customers who are likely to leave the bank.

<h4>Insight:</h4>

For customer churn prediction, identifying at-risk customers is more important than maximizing overall accuracy. The Decision Tree model provides better business value because it captures a much larger proportion of actual churn customers, enabling more effective retention campaigns.

<h1>Random Forest<h1>

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, rf_pred)*100,'\n')

print(confusion_matrix(y_test, rf_pred),'\n')

print(classification_report(y_test, rf_pred))
```

<h4>Observation:</h4>

Random Forest outperformed both Logistic Regression and Decision Tree by achieving the highest accuracy (86.4%), precision (78%), and F1-score (58%). Although its recall (46%) was slightly lower than Decision Tree, it provided the best overall balance between identifying churn customers and minimizing false predictions.

<h4>Insight:</h4>

Random Forest is the most suitable model for customer churn prediction because it accurately identifies at-risk customers while maintaining high overall prediction performance. This enables the bank to target retention campaigns more effectively and reduce customer loss with fewer unnecessary offers.
